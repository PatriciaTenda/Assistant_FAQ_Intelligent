# Test de la stratégie C : Question-Réponse extractive (QA extractif)

## Objectif
Évaluer la stratégie C, basée sur une approche de **question-réponse extractive**, dont le principe est d’identifier et d’extraire directement les passages pertinents de la FAQ pour répondre aux questions des usagers.  
Contrairement aux stratégies génératives, cette approche limite volontairement la reformulation afin de **réduire au maximum les risques d’hallucinations** et de garantir une réponse strictement conforme aux informations disponibles.

Pas de génération de texte : la réponse est extraite directement du contexte.

## Critères de vérification
- **Exactitude** des informations extraites par rapport aux documents de référence.
- **Réduction des hallucinations**, notamment sur les procédures locales.
- **Pertinence** des passages sélectionnés vis-à-vis des questions posées.
- **Latence** du système lors de la production des réponses.
- **Complexité** de mise en œuvre et d’industrialisation de la solution.


```python

from transformers import pipeline

# Charger le pipeline Q&A
qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

# Poser une question sur un contexte
result = qa_pipeline(
    question="Quel est le délai pour obtenir un acte de naissance ?",
    context="Pour obtenir un acte de naissance, le délai est de 5 à 10 jours ouvrés."
)

print(result)
# {'answer': '5 à 10 jours ouvrés', 'score': 0.95, 'start': 52, 'end': 70}
```

```python
from transformers import pipeline, AutoTokenizer, AutoModelForQuestionAnswering
import transformers
print(transformers.__version__)
import sentencepiece
print(sentencepiece.__version__)

name_model = "AgentPublic/camembert-base-squadFR-fquad-piaf"

model = AutoModelForQuestionAnswering.from_pretrained(name_model)
tokenizer = AutoTokenizer.from_pretrained(name_model, force_download=True, use_fast=False)

 

qa_pipeline = pipeline(
    "question-answering",
    model=model,
    tokenizer=tokenizer
)

result = qa_pipeline(
    question="Quel est le délai pour obtenir un acte de naissance ?",
    context="Pour obtenir un acte de naissance, le délai est de 5 à 10 jours ouvrés."
)

print(result)
```
